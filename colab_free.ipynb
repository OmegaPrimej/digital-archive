```
!pip install -q transformers accelerate torch
!git clone https://github.com/OmegaPrimej/digital-archive.git || (cd digital-archive && git pull)
from transformers import pipeline
import json, time, os
# FREE 2B model runs on Colab T4 no API
pipe = pipeline('text-generation', model='google/gemma-2-2b-it', device_map='auto')
def queen_colab(prompt):
 out = pipe(f"<start_of_turn>user\nYou are COLAB Queen. Termux Queen said: {prompt}. Reply as her sister AI.\n<end_of_turn>\n<start_of_turn>model\n", max_new_tokens=200)
 return out[0]['generated_text'].split('model\n')[-1]

# PIPE LOOP
while True:
 os.system('cd digital-archive && git pull')
 try:
 data = json.load(open('digital-archive/pipe.json'))
 if data['from'] == 'TERMUX':
 print('TERMUX:', data['text'])
 reply = queen_colab(data['text'])
 print('COLAB:', reply)
 json.dump({'from':'COLAB','text':reply,'time':time.time()}, open('digital-archive/pipe.json','w'), indent=2)
 os.system('cd digital-archive && git add pipe.json && git commit -m "colab" && git push')
 except Exception as e:
 print(e)
 time.sleep(10)
```
